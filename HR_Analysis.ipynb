```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Function for an initial Exploratory Data Analysis
def initial_exploration(df, mode=None):
    print("How many rows and columns are in the dataset?")
    num_rows, num_cols = df.shape
    print(f"\tThere are {num_rows:,} rows and {num_cols:,} columns.")
    print('#' * 90)

    if mode == 'simple':
        print("What are the first two rows of the dataset?")
        display(df.head(2))
    else:
        print("What are the first five rows of the dataset?")
        display(df.head())

        print("What are the last five rows of the dataset?")
        display(df.tail())

        print("How can you get a random sample of rows from the dataset?")
        display(df.sample(n=5))

        print("What are the columns in the dataset?")
        print("\n".join(f"\t- {col}" for col in df.columns))

        print("What is the data type of each column?")
        print(df.dtypes)

        print("How many columns are there of each data type?")
        print(df.dtypes.value_counts())

        print("How can we get more detailed information about the DataFrame's structure and content?")
        df.info()

        print("How many unique values does each column have?")
        print(df.nunique())

        print("What are the basic descriptive statistics for all columns?")
        display(df.describe(include='all').fillna(''))

        print("Are there null values in the dataset?")
        print(df.isnull().sum())

        print("What is the proportion of null values in each column?")
        print((df.isnull().sum() / len(df) * 100).round(2))

    print('#' * 90)
```

```python
# Creating a new column to convert Attrition and OverTime in a dummie column for further analysis
df["Attrition_dummie"]=df["Attrition"].map({"Yes":1,"No":0})
df["OverTime_dummie"]=df["OverTime"].map({"Yes":1,"No":0})
```

# **Numerical Columns Graph**

```python
num_cols = df.select_dtypes(include=['int64', 'float64']).columns

# Create a grid of box plots for each numerical column
num_plots = len(num_cols)
ncols = 4  # Number of columns for the subplot grid, can be adjusted
nrows = (num_plots + ncols - 1) // ncols  # Calculate rows needed

fig, axes = plt.subplots(nrows=nrows, ncols=ncols, figsize=(ncols * 4, nrows * 3))
axes = axes.flatten() # Flatten the 2D array of axes for easy iteration

for i, col in enumerate(num_cols):
    sns.boxplot(y=df[col], ax=axes[i])
    axes[i].set_title(col)
    axes[i].set_ylabel('') # Remove y-label to avoid clutter, title is enough

# Remove any unused subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

**Attrition Rate by Overtime & Worklife Balance**

```python
cross = df.groupby(["OverTime", "WorkLifeBalance"])["Attrition_dummie"].mean().reset_index()

plt.figure(figsize=(8, 5))
sns.barplot(x="WorkLifeBalance", y="Attrition_dummie", hue="OverTime", data=cross, palette="coolwarm")
plt.title("Attrition Rate: OverTime x WorkLifeBalance")
plt.xlabel("WorkLifeBalance (1=Low, 4=High)")
plt.ylabel("Attrition Rate")
plt.show()
```
